```python
from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.6-flash",
    temperature=0.1,
    max_retries=5 
)

llm.call("What is the capital of France?")
```

```python
original_email = """
looping in Priya.TAS and PRX updates are in the deck. ETA for SDS integration is Friday.
Let's sysn up tomorrow if SYNCBOT allows. Ping me if any blockers.
"""

from crewai.tools import BaseTool

class ReplaceJargonsTool(BaseTool):
    name:str = "Replace Jargons"
    description: str = "Replace jargons with more specific terms"
    
    def _run(self, email: str) ->str:
        replacements ={
            "PRX": "Project Phoenix (internal AI revamp project)",
            "TAS": "Technical Architecture Stack",
            "DBX": "Client Database cluster",
            "SDS": "Smart Data Syncer",
            "SYNCBOT": "internal standup assistant bot",
            "WIP": "work in progress",
            "POC": "proof of concept",
            "ping": "Reach out"
        }
        
        suggestions =[]
        email_lower = email.lower()
        for jargon, replacement in replacements.items():
            if jargon.lower() in email_lower:
                suggestions.append(f"Replace '{jargon}' with '{replacement}'")
        return "\n".join(suggestions) if suggestions else "No jargons found to replace."

jt = ReplaceJargonsTool() ###JT is our tool here.
jt.run(original_email)
```

```python
from crewai import Agent, Task, Crew

email_assistant = Agent(
    role="You are an email agent that can read and respond to emails.",
    goal="Your goal is to read emails and respond to them appropriately.",
    backstory="You are a highly skilled email agent with experience in handling various types of emails.",
    tools=[jt],
    verbose=True,
    llm=llm
)

email_task = Task(
    description=f"""Read the following email and respond to it in a professional and polished version.
    Expand abbreviations:
    '''{original_email}'''
    """,
    agent=email_assistant,
    expected_output="A professional written email with proper formatting and content."
)

crew = Crew(
    agents=[email_assistant],
    tasks=[email_task],
    verbose=True   
)

result = await crew.kickoff_async()
print(result)
```
